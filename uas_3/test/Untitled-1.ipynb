```python
from datetime import datetime, timedelta

# ===============================
# DATA & TARIF
# ===============================
TARIF = {
    "mobil": 5000,
    "motor": 2000,
    "lain": 3000
}

riwayat = [
    {"plat": "111", "jenis": "motor", "merk": "Vario", "masuk": "2026-01-15 09:01:50", "keluar": "2026-01-15 10:01:50"},
    {"plat": "101", "jenis": "motor", "merk": "Beat", "masuk": "2026-01-15 13:01:50", "keluar": "2026-01-15 14:01:50"},
    {"plat": "222", "jenis": "mobil", "merk": "Avanza", "masuk": "2026-01-15 09:02:50", "keluar": "2026-01-15 10:02:50"},
    {"plat": "201", "jenis": "mobil", "merk": "Pick Up", "masuk": "2026-01-15 09:03:50", "keluar": "2026-01-15 10:03:50"},
    {"plat": "102", "jenis": "motor", "merk": "Scoopy", "masuk": "2026-01-15 15:01:50", "keluar": "2026-01-15 16:01:50"},
    {"plat": "103", "jenis": "motor", "merk": "Supra", "masuk": "2026-01-15 15:04:50", "keluar": "2026-01-15 16:04:50"},
    {"plat": "202", "jenis": "mobil", "merk": "Suzuki", "masuk": "2026-01-15 15:10:50", "keluar": "2026-01-15 16:10:50"},
    {"plat": "208", "jenis": "mobil", "merk": "Suzuki", "masuk": "2026-01-15 15:15:50", "keluar": "2026-01-15 16:15:50"},
    {"plat": "209", "jenis": "mobil", "merk": "Suzuki", "masuk": "2026-01-15 15:20:50", "keluar": "2026-01-15 16:20:50"},
    {"plat": "210", "jenis": "mobil", "merk": "Suzuki", "masuk": "2026-01-15 15:40:50", "keluar": "2026-01-15 16:40:50"},
    {"plat": "211", "jenis": "mobil", "merk": "Suzuki", "masuk": "2026-01-15 17:10:50", "keluar": "2026-01-15 18:10:50"},
    {"plat": "212", "jenis": "mobil", "merk": "Suzuki", "masuk": "2026-01-15 06:10:50", "keluar": "2026-01-15 07:10:50"},
    {"plat": "213", "jenis": "mobil", "merk": "Suzuki", "masuk": "2026-01-15 07:10:50", "keluar": "2026-01-15 08:10:50"},
    {"plat": "113", "jenis": "motor", "merk": "Suzuki", "masuk": "2026-01-15 15:10:50", "keluar": "2026-01-15 16:10:50"},
    {"plat": "114", "jenis": "motor", "merk": "Yamaha", "masuk": "2026-01-15 08:00:00", "keluar": "2026-01-15 09:00:00"},
    {"plat": "115", "jenis": "motor", "merk": "Honda", "masuk": "2026-01-15 10:00:00", "keluar": "2026-01-15 11:00:00"},
    {"plat": "214", "jenis": "mobil", "merk": "Toyota", "masuk": "2026-01-15 11:30:00", "keluar": "2026-01-15 12:30:00"},
    {"plat": "215", "jenis": "mobil", "merk": "Daihatsu", "masuk": "2026-01-15 12:45:00", "keluar": "2026-01-15 13:45:00"},
    {"plat": "116", "jenis": "motor", "merk": "Kawasaki", "masuk": "2026-01-15 14:00:00", "keluar": "2026-01-15 15:00:00"},
    {"plat": "216", "jenis": "mobil", "merk": "Mitsubishi", "masuk": "2026-01-15 16:30:00", "keluar": "2026-01-15 17:30:00"},
]

# ===============================
# FUNGSI ANALISIS
# ===============================
def hitung_rekap(riwayat, mulai):
    rekap = {
        "motor": {"jumlah": 0, "pendapatan": 0},
        "mobil": {"jumlah": 0, "pendapatan": 0},
        "lain": {"jumlah": 0, "pendapatan": 0},
    }

    for r in riwayat:
        masuk = datetime.strptime(r["masuk"], "%Y-%m-%d %H:%M:%S")
        keluar = datetime.strptime(r["keluar"], "%Y-%m-%d %H:%M:%S")

        if keluar < mulai:
            continue

        jam = (keluar - masuk).total_seconds() / 3600
        jam = max(1, int(jam))

        tarif = TARIF.get(r["jenis"], TARIF["lain"])
        pendapatan = jam * tarif

        rekap[r["jenis"]]["jumlah"] += 1
        rekap[r["jenis"]]["pendapatan"] += pendapatan

    return rekap


def tampilkan(judul, rekap):
    print(f"\n=== {judul} ===")
    total = 0
    for jenis, data in rekap.items():
        if data["jumlah"] > 0:
            total += data["pendapatan"]
            print(
                f"{jenis.upper():<6} : "
                f"{data['jumlah']} kendaraan | "
                f"Rp {data['pendapatan']:,}".replace(",", ".")
            )
    print(f"TOTAL : Rp {total:,}".replace(",", "."))


# ===============================
# PERIODE ANALISIS (AMBIL DARI DATA)
# ===============================
now = max(
    datetime.strptime(r["keluar"], "%Y-%m-%d %H:%M:%S")
    for r in riwayat
)

hari_ini = datetime(now.year, now.month, now.day)
minggu_ini = now - timedelta(days=7)
bulan_ini = datetime(now.year, now.month, 1)
tahun_ini = datetime(now.year, 1, 1)

# ===============================
# OUTPUT
# ===============================
tampilkan("HARI INI", hitung_rekap(riwayat, hari_ini))
tampilkan("1 MINGGU TERAKHIR", hitung_rekap(riwayat, minggu_ini))
tampilkan("BULAN INI", hitung_rekap(riwayat, bulan_ini))
tampilkan("TAHUN INI", hitung_rekap(riwayat, tahun_ini))
```
